## What the book has shown

The book began with a wall and ends with a map of the ways around it.
The wall, the exponential growth of the many-body Hilbert space, was
stated in Chapter 1 and quantified in
Chapter 3, where the four models that run through the
book, Lipkin, pairing, Hubbard and Heisenberg, were solved exactly at
small size and approximately by Hartree--Fock and coupled cluster, and
where the unitary coupled-cluster ansatz appeared as the first object
whose evaluation is natural on a quantum computer and unnatural on a
classical one. Chapter 4 turned the second-quantized
Hamiltonians into sums of Pauli strings and showed that the cost of a
model on a quantum computer is measured by three numbers, qubits,
strings and weight, each of which a symmetry can reduce; the Lipkin,
pairing and Heisenberg models all became four-qubit problems of
two-qubit strings, and the eight-qubit Hubbard ring kept the parity
strings that are the price of fermions. From the Pauli sum three
circuits followed mechanically, the Pauli rotation, the Trotter step
and the measurement of an energy, and everything in the later chapters
was built from them.

The algorithms then divided along a line that the book has drawn many
times. On one side stand the methods that need a deep coherent
circuit and give a guarantee: Hamiltonian simulation in
Chapter 5, whose Trotter error is controlled and
whose cost is polynomial in the system size and the evolution time;
phase estimation in Chapter 6, which converts that
simulated evolution into an eigenvalue with a precision that doubles
with every control qubit, and whose application to the four models
showed the accuracy governed entirely by the register and the time
step and not by the difficulty of the system; and the rodeo algorithm
and its relatives in Chapter 8, which use the same
controlled evolutions to project an eigenstate out of a trial state
with one reusable ancilla, an error that falls exponentially with the
number of cycles, and a probabilistic, heralded outcome whose success
rate is the overlap of the trial state, which resolution refinement
was introduced to make large; compiled to gates, the pairing
calculation of that chapter came to $10^4$ to $10^5$ two-qubit gates
on five to nine qubits. On the other side stand the variational methods
of Chapter 7, which need only shallow circuits and many
repetitions, give an upper bound and nothing more, and whose quality
was seen to be a property of the ansatz rather than of the algorithm:
a hardware-efficient circuit converged smoothly to the wrong
particle-number sector of the pairing and Hubbard Hamiltonians, while
circuits built from the physics of each model were exact with a
handful of parameters, and the adaptive version, given a pool
complete under the symmetries of the problem, found those circuits by
itself. The two sides need each other, since the variational state
with its overlap of $0.9998$ is what makes a phase-estimation run
succeed on the first attempt, and the phase-estimation run is what
certifies the variational state; that combination is the standard
plan for the fault-tolerant era, and the small calculations of this
book are its rehearsal.

Chapter 9 explained why the fault-tolerant era is needed
at all and how it will be reached: the error models, the codes that
detect and correct them by measuring syndromes rather than states,
the surface code whose threshold has now been crossed on hardware, and
the mitigation techniques that, in the meantime, trade repetitions for
accuracy on the variational side. Chapter 10 took the
whole apparatus to the problem that motivates much of the effort in
nuclear and particle physics, a lattice gauge theory whose real-time
and finite-density physics is closed to classical Monte Carlo by the
sign problem, and showed on the Schwinger model that the mapping, the
circuits and the measurements of the earlier chapters extend to gauge
fields, and at what cost. What the book has not shown is a single
result beyond the reach of a classical computer. Every table in it
was produced on a simulator of at most eight qubits, checked against
exact diagonalization, and chosen to be checkable; that was the point,
and it is also the honest state of the field for the problems of
many-body physics, where the calculations that have been run on
hardware are, with a few exceptions in dynamics, still ones a laptop
can verify.


## Where the difficulty went

A theme has run through the book and deserves to be stated once more
as the conclusion it is. Quantum computers do not remove the
difficulty of the many-body problem; they move it, and the different
algorithms move it to different places. In phase estimation the
difficulty lives in state preparation, since the algorithm returns the
energy of whatever eigenstate the trial state overlaps with, and in
circuit depth, since every bit of precision doubles the number of
applications of the evolution operator. In the variational methods it
lives in the design of the ansatz, which decides whether the answer is
close, and in the measurement count, which grows as the square of the
inverse accuracy and of the Pauli norm of the Hamiltonian, and which
no proposal so far reduces below that scaling. In Hamiltonian
simulation it lives in the Trotter error, or in the ancilla overhead
of the methods that replace Trotterization, and in the extraction of
observables from the evolved state. In every case there is a quantity,
the overlap, the measurement budget, the depth, that is polynomial for
the small systems of this book and whose behaviour at large size is
either unknown or known to be unfavourable in the worst case. The
question of whether quantum computers offer an exponential advantage
for ground-state problems reduces, as Chapter 6 discussed,
to whether the overlap of an efficiently preparable state with the
ground state decays polynomially or exponentially with the system size
for the systems of interest, and the evidence, reviewed critically in
[Lee2023], is that for the molecules and materials studied so far
the answer is not the one the field had hoped for. For nuclei, whose
ground states are often well approximated by a small number of
configurations in a suitably chosen basis, and for real-time dynamics,
where the advantage rests on Lloyd's theorem and not on an overlap,
the prospects are better, and it is no accident that the case study of
Chapter 10 is a problem in dynamics.


## The hardware

Every algorithm in the book was run on a simulator, and it is worth
closing with a sober account of the machines it was written for. Four
platforms carry most of the effort, and they differ in the resources
that matter for many-body simulation, the number of qubits, the
fidelity and speed of the two-qubit gates, the connectivity of the
qubits and the coherence time, in ways that make each the natural
home of a different class of problem [Preskill2018,Daley2022].
Superconducting circuits, the platform of the largest commercial
efforts, have hundreds of qubits on a chip with nearest-neighbour
connectivity on a planar lattice, gate times of tens of nanoseconds,
two-qubit gate errors that have fallen to a few parts in a thousand,
and coherence times of a few hundred microseconds; the utility-scale
dynamics experiments and the below-threshold surface-code experiments
cited in Chapter 1 were performed on them, and their
planar connectivity is a direct match to the surface code and to
lattice models in two dimensions, while the parity strings of a
fermionic mapping must be routed through swaps. Trapped ions have
fewer qubits per trap, tens rather than hundreds, but all-to-all
connectivity within a trap, the highest two-qubit gate fidelities
demonstrated on any platform, and coherence times of seconds; their
gates are slower by orders of magnitude, and their natural strength is
in problems with long-range interactions and in the deep circuits with
few qubits that phase estimation on small systems requires. Neutral
atoms in optical tweezers combine hundreds to thousands of qubits,
reconfigurable connectivity by moving the atoms, and, through the
Rydberg blockade, entangling gates that have improved rapidly; the
first logical-qubit experiments at scale were performed on them
[Bluvstein2024], and the same platform, operated in an analog
rather than a digital mode, is a quantum simulator of spin models in
its own right. Photonic and spin-qubit platforms are less mature for
the problems of this book and are not discussed here.

Two remarks apply to all of them. The first is that the numbers that
matter for a many-body calculation are the product of the two-qubit
gate error and the number of two-qubit gates in the circuit. A Trotter
step of the eight-qubit Hubbard ring of Chapter 4
costs a few dozen CNOT gates, a phase-estimation run with ten control
qubits costs thousands of Trotter steps, and a gate error of $10^{-3}$
gives an error budget of a few hundred gates before the output is
noise; the arithmetic is the reason the variational methods of
Chapter 7 exist and the reason error correction is not
optional. The second is that the demonstrations of error correction
below threshold, on superconducting and on neutral-atom platforms,
change the character of the projection. A logical qubit whose error
falls as the code distance grows means that the depth budget is no
longer fixed by the physics of the qubit but by the number of physical
qubits one is willing to spend per logical one, of the order of a
thousand for the surface code at useful error rates; the resource
estimates for chemically relevant molecules and for nuclei in realistic
model spaces, which run to millions of physical qubits and days of
runtime, are the targets of the next decade of engineering, and they
are being revised downwards by better algorithms as fast as the
hardware improves.


## Beyond the gate model

This book has been about the gate model, in which a many-body problem
is compiled into a sequence of discrete unitaries on qubits, because
the gate model is universal, because its errors can be corrected, and
because its algorithms have guarantees that can be stated as theorems.
It is not the only way to use a quantum system to study another. The
oldest is the *analog quantum simulator*, closest to Feynman's
original proposal, in which a controllable system, ultracold atoms in
an optical lattice, trapped ions with tunable spin-spin couplings,
Rydberg atoms in tweezer arrays, superconducting circuits operated as
coupled oscillators, is engineered to have the Hamiltonian one wants
and then simply observed [Bloch2012,Georgescu2014]. The Hubbard
model of Chapter 3 has been realized this way with
fermionic atoms at temperatures now approaching the scale of the
antiferromagnetic exchange, the Heisenberg and transverse-field Ising
models with ions and Rydberg atoms at hundreds of spins, and lattice
gauge theories of the kind studied in Chapter 10 in the
first small instances. Analog simulators have no error correction and
no guarantee of accuracy, and their observables are limited to what
the apparatus can measure; what they have is size and speed, and for
the equilibrium and dynamical properties of lattice models in two
dimensions they are ahead of any gate-model calculation and likely to
remain so for some time. The relation between the two approaches is
the relation between an experiment and a computation, and the field
needs both.

*Adiabatic quantum computation* and *quantum annealing*
occupy the ground between. The adiabatic theorem that motivated the
QAOA ansatz of Chapter 7 can be run directly: prepare the
ground state of a simple Hamiltonian, deform it slowly into the one of
interest, and read out the result, with a running time set by the
inverse square of the minimal gap along the path. As a model of
computation it is equivalent to the gate model; as a practice, on the
special-purpose annealers that implement transverse-field Ising
Hamiltonians on thousands of qubits with limited coherence, it is a
heuristic optimizer whose advantage over classical methods on any
problem remains unestablished, and whose relevance to many-body
physics lies mainly in its use as a sampler of classical spin-glass
and, with some care, quantum spin-model configurations. The
*hybrid* methods are the most recent and, for the near term, the
most promising. Every variational algorithm of Chapter 7 is
already hybrid, a quantum circuit inside a classical optimizer, and the
pattern generalizes: a quantum computer can serve as the impurity
solver inside a dynamical mean-field or embedding calculation, as the
evaluator of a few hard matrix elements inside a classical
configuration-interaction or coupled-cluster calculation, as the
generator of samples for a classical tensor-network or
neural-network state, or as the provider of a trial state that a
classical method refines. In each of these the quantum resource is
spent on the part of the problem where classical methods fail and the
classical resource on everything else, which is where the exponential
lives and where it does not, respectively. The quantum-machine-learning
models glanced at in Chapter 7 belong here too, as
parametrized circuits trained on data, and the question of whether
they offer anything that a classical model does not is open in the
same way, and for the same reasons, as the question for variational
eigensolvers.


## Open problems

The problems that remain are of three kinds, and it is useful to keep
them apart, since they call for different people. The first kind is
hardware: more qubits, better gates, longer coherence, faster readout,
and the engineering of error correction from a demonstration into a
platform. This is not a problem many-body theorists will solve, but it
is one they should watch, because the algorithms that will run first
on error-corrected machines are those with the smallest constant
factors, and those factors are set by choices, of encoding, of Trotter
ordering, of symmetry sector, that are theirs to make. The second
kind is algorithmic, and here the many-body theorist is the expert.
State preparation is the largest of them: the overlap problem of
Chapter 6 is a question about the structure of physical
ground states, about which many-body theory knows a great deal and
quantum information theory little, and the design of trial states
with provably large overlap, from mean-field theory, from coupled
cluster, from tensor networks, from adiabatic paths, is where the two
fields most need each other. Measurement is the second: the
$\epsilon^{-2}$ scaling of the variational methods is a statement
about sample means, but the prefactor, the Pauli norm of the
Hamiltonian, depends on the basis and the mapping and has been reduced
by orders of magnitude for molecules by better choices, and the
corresponding work for nuclear and lattice Hamiltonians is in its
infancy. The third is the ansatz, whether hand-designed, adaptive or
learned, and the barren-plateau obstruction that limits all of them:
the escape routes, symmetry-preserving circuits, physically motivated
initializations and problem-specific pools, are exactly the places
where physical insight enters, and where the failures documented in
Chapter 7 turn into a research programme. Beyond these,
the extraction of observables other than the energy, response
functions, transition rates, scattering amplitudes and the real-time
correlators of Chapter 10, is where the advantage of
Hamiltonian simulation should first become visible, and where the
methods, including the resolvent techniques of Chapter 6,
are least developed.

The third kind of problem is one of judgement. Classical many-body
methods have not stood still: tensor networks in two dimensions,
neural-network quantum states, and the steady improvement of
coupled-cluster, Monte Carlo and configuration-interaction methods
have moved the boundary of what a classical computer can do every
year, and every claim of quantum advantage is a claim about that
moving boundary. The discipline the field needs is the one this book
has tried to practise: to compare against the best classical method on
the same problem, to state the error and its source, to distinguish a
theorem from a heuristic and a demonstration from a result, and to say
when something is not known. The problems of many-body physics are
hard enough that no tool will be wasted, and the reader who has
reached this page has the two toolkits, the classical one of
Chapter 3 and the quantum one of the chapters that
followed, in the same hands. That is, in the end, what the book was
for. The interesting work is in using both.
