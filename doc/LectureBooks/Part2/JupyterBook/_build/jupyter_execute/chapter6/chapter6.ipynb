## Scope of this chapter

This chapter is a placeholder for material to be written. It is kept
in the book so that the structure announced in
Section *The plan of the book* is visible, and so that the labels and the
program folder (`BookPrograms/chapter6/`) are reserved.

Reinforcement learning is the third kind of learning named in
Section *What machine learning is*. An *agent* observes the state $s_t$ of an
*environment*, chooses an action $a_t$ according to a
*policy* $\pi(a\vert s;\vec{\theta})$, receives a reward $r_t$,
and must adjust $\vec{\theta}$ so as to maximize the expected
cumulative reward. There are no labeled samples: the data are
generated by the agent's own actions, and the central quantities are
the value function of a state, the action-value function, and the
policy gradient, which has the same parameter-shift form as the
gradients of Chapter 4 when the policy is a circuit.

Quantum reinforcement learning comes in three forms, and the chapter
will treat them in this order.

- *Quantum agents for classical environments*: the policy or
   the action-value function is a parametrized circuit of the kind of
   Chapter 4, with the state encoded as data, trained by
   policy-gradient or $Q$-learning updates; the questions are the ones of
   Section *Variational circuits reconsidered*, trainability, classical surrogates
   and total cost, in a setting where the data are generated online.
- *Classical agents for quantum environments*: the
   environment is a quantum system to be controlled, prepared or
   measured, the state is a description of it or of the measurement
   record, and the reward is a fidelity or an estimation accuracy. This
   is the form closest to the many-body and quantum-control interests of
   this book and the one in which, following the argument of
   Section *Variational circuits reconsidered*, we expect the most durable results:
   reinforcement learning for state preparation, gate design, error
   correction and adaptive experiment design.
- *Quantum agents for quantum environments*: agent and
   environment are both quantum and may be entangled, which is the
   setting of the quantum-enhanced learning-from-experiments results of
   Sections *Variational circuits reconsidered* and *A priority list for the next years*.

The intended sections are: Markov decision processes and the Bellman
equations; value-based and policy-based classical algorithms
($Q$-learning, policy gradients, actor--critic) summarized from the
reinforcement-learning part of the machine-learning text [MLbook]
and implemented in `common/`; the variational circuit as a policy,
with its gradient by the parameter-shift rule and its connection to
the Fourier picture of Section *The output as a Fourier series*; reinforcement
learning for quantum control, with the state-preparation and
gate-design problems of the companion volume [QCbook] as
environments; adaptive measurement and experiment design; and a
closing comparison with the classical baselines. Each section will
follow the pattern of the preceding chapters, with every number
produced by a program and both the classical and the quantum version
run on the same problem.
