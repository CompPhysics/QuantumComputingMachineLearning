## What the book has shown

The four model chapters have built, from the one hybrid model of
Eq. (1.1), a kernel method, an optimization algorithm, a
neural network with a physics-informed variant, and a generative
model, each with its classical counterpart beside it and each tested
exactly on problems small enough to allow it. It is worth collecting
what those exact computations showed, model by model, because the
pattern is the evidence on which the rest of this chapter rests.

**The quantum support-vector machine.** 
A new kernel, not a new algorithm. The angle-encoding kernel is a
product of cosines and is evaluated classically in $\bigO(p)$
operations; the entangling feature map of Havlíček and
coworkers is not known to be, and on the data set built from its own
structure it is exact where the Gaussian kernel is not
(Chapter 2). On the data sets one actually meets (moons,
circles, XOR, Iris), the quantum kernels matched the classical ones and
did not exceed them, and the estimated kernel needs a number of shots
that grows with the size of the Gram matrix. The advantage, where
there is one, is in the data, as Huang and coworkers
[Huang2021] say, and the model's job is to not stand in its way.

**The quantum approximate optimization algorithm.** 
A variational eigensolver for a classical spin glass, whose ansatz
is Trotterized adiabatic evolution and whose parameters are trained
by the same loop as every later model (Chapter 3). At
the depths we simulated it reaches the optimum of the smallest
instances, an edge and a ring, but not of a six-vertex graph with a
triangle; its landscape has the structure that warm starts exploit,
and the per-term shift rule makes its gradients exact; what it does
not do is beat the classical heuristics that solve these instances
instantly, and the portfolio and clustering applications showed the
model as a flexible optimizer rather than as a faster one.

**The quantum neural network.** 
A trigonometric polynomial in the encoded data whose frequencies are
fixed by the encoding and whose coefficients are set by the trainable
angles (Chapter 4). This single fact explains every
experiment: the regression that an 18-parameter circuit fits at the
noise floor and a 321-parameter network fits no better, the
classifiers that match the kernel machines, the physics-informed
circuits that solve a Poisson or a wave equation to machine precision
when the solution lies in their function class and not otherwise. The
barren plateaus of random circuits were reproduced with their
exponential laws, and the natural gradient, imaginary-time evolution
and stochastic reconfiguration were shown to be one equation.

**The quantum Boltzmann machine.** 
Statistical mechanics as a generative model (Chapter 5).
On classical data the likelihood drove the transverse fields to zero
and the model fell back on its classical limit, so that the training
rule of the literature, the gradient of a Golden--Thompson bound, was
good enough because the exact gradient wanted the same thing. On
quantum data the quantum relative entropy was convex, Hamiltonians
were recovered exactly from their thermal states, and the operator
content of the model, non-commuting terms that no classical energy
function can even represent, decided what could be learned.

The pattern is the one announced in Section *Variational circuits reconsidered*.
The quantum models work as their theory says. On the classical data
sets a textbook can hold, their gains over classical methods are
gains of parameter economy or of structure rather than of accuracy,
and in every case we could name the classical surrogate: the product
of cosines, the single-layer product form, the classical Boltzmann
machine. On quantum data (a Hamiltonian's ground or thermal state, a
kernel between quantum states, the operator content of a model
Hamiltonian) there is no classical competitor to begin with, and the
quantum model is not a replacement for a classical one but the only
model there is.


## The reassessment, with the evidence in hand

Section *Variational circuits reconsidered* stated three connected difficulties of
variational quantum machine learning and promised that the chapters
would meet each in a concrete form. They did. *Trainability*: the
gradient variances of Table 4.2 fall as $2^{-n}$ and
$4^{-n}$ for the local and global costs, the kernel values of expressive
feature maps concentrate exponentially (Section *Perspectives*),
and the shot-noise study of Section *Kernels from a finite number of shots* shows what finite
statistics do to a kernel matrix. *Classical competition*: every
model that trained well in this book had a classical surrogate we
could write down, and the published results on the link between
provable absence of barren plateaus and classical simulability
[Cerezo2025] and on classical surrogates of
quantum convolutional networks [Bermejo2026] say that this is not
an artifact of our small examples. *Total cost*: the kernels of
Chapter 2 need $\bigO(N^2)$ circuit runs of $S$ shots each, the
gradients of Chapter 4 need $2P$ circuit evaluations per sample per
step, and the Boltzmann machine of Chapter 5 needs a Gibbs state prepared
anew for every gradient step; none of these costs appears in the
classical competitor.

None of this is a theorem against variational learning. Avoiding
barren plateaus does not universally imply classical simulability;
the connection has been established case by case, for many commonly
proposed architectures, and the surrogate sometimes requires a quantum
data-acquisition stage that a classical computer cannot replace. The
surrogate of a quantum convolutional network may still need quantum
measurements of its input, and what it removes is the repeated quantum
training loop. These results are serious challenges to particular
designs. What they change is the burden of proof: a proposed quantum
learning model must now say which quantum resource it uses and what
that resource buys, and "it trains" and "it generalizes" are no
longer answers.


## A priority list for the next years

Table 7.1 gives, for the next three to five years,
the directions we find promising, in our order of priority. It is a
judgment rather than a consensus, and each entry carries the
qualification that we think decides it.

| Approach | Why we find it promising | Main qualification |
|---|---|---|
| Learning quantum states, processes and dynamics | Quantum inputs avoid the classical data-loading bottleneck; coherent access and entangled measurements can give sample advantages | The advantages depend on the task and on the experimental access permitted |
| Quantum measurements followed by classical learning | Extract informative features once, then train classical models; often more economical than repeatedly optimizing a circuit | Some tasks need quantum memory or collective measurements that an ordinary measurement record cannot preserve |
| Structured variational models | Symmetry, locality, conservation laws and physical dynamics provide an inductive bias | Trainability and good predictive performance do not by themselves establish a quantum advantage |
| Quantum reservoir computing and fixed feature maps | Let quantum dynamics generate features and train mainly the classical readout; little optimization overhead | Measurement cost and strong classical reservoir baselines remain decisive |
| Fault-tolerant learning primitives | Coherent algorithms open possibilities beyond shallow noisy circuits | Practical value depends on logical hardware, data access and full resource estimates |

*Table 7.1: Directions in quantum machine learning for the near term, in
our order of priority, with the reason for the ranking and
the qualification that will decide each.*

**Learning quantum states, processes and dynamics.** 
There is concrete evidence behind the first direction. The theory of
learning from experiments [Huang2022experiments] shows an
exponential separation in the number of experiments between an agent
that can store and jointly measure copies of a quantum state and one
that measures each copy and reasons classically, and the photonic
experiment of Liu and coworkers [Liu2025photonic] realized, using
entanglement, such a separation for the learning of bosonic
displacement processes. This is evidence for learning quantum processes, not for
a general quantum advantage on ordinary classification data, and the
Hamiltonian learning of Section *Learning quantum states* is its simplest
textbook instance: a convex problem whose data are expectation values
of the state to be learned. The sample-complexity results of Anshu and
coworkers [Anshu2021] for Gibbs states, and the provably efficient
classical learning of ground-state properties from measurement data
[Huang2022provably], set the frame in which such claims should be
made.

**Quantum measurements followed by classical learning.** 
The second direction follows directly from the classical-surrogate
results: the essential quantum contribution may be acquiring
information, with the useful learning done classically. Classical
shadows [Huang2020shadows] are the tool, a randomized measurement
protocol from which many observables of a state can be predicted from
few copies, and the surrogate of the quantum convolutional network
[Bermejo2026] is the example: the measurements of the input are
quantum, the training is not. The qualification is that some tasks
need a quantum memory or collective measurements that no record of
single-copy outcomes preserves, which is exactly the resource that
the first direction uses.

**Structured variational models.** 
Variational circuits retain a credible role here. Circuits that
respect the symmetries of a learning problem
[Larocca2022,Meyer2023symmetry] encode its structure and improve
generalization in the examples studied, and for many-body physics the
natural structure is particle number, spin, spatial symmetry and the
observables of interest, which is the content of the physically
motivated ansätze of the companion volume [QCbook]. The
qualification is the one of Section *The reassessment, with the evidence in hand*: a
structured circuit that trains and predicts well has not thereby been
shown to outperform a tensor network or a classical network with the
same symmetries built in.

**Quantum reservoir computing and fixed feature maps.** 
The fourth direction removes the optimization loop. A quantum system
with fixed dynamics is driven by the input, its measured observables
are the features, and a linear readout is trained classically
[Fujii2017]; the circuit-QED demonstration of Carles and coworkers
[Carles2026], with a transmon--resonator system generating
nonlinear features and no quantum gradient-training loop, is a recent
proof of concept. The kernel methods of Chapter 2 are the
same idea with the readout a support-vector machine. Whether a quantum
reservoir beats a classical one at the same measurement budget is the
decisive, and open, question.

**Fault-tolerant learning primitives.** 
The fifth direction is the one of the original speedup claims of
Section *Why quantum?*, read with Aaronson's fine print
[Aaronson2015] and Tang's dequantizations [Tang2019] in mind.
Coherent algorithms (phase estimation, amplitude estimation, the
linear-system algorithm and quantum principal-component analysis
[Lloyd2014qpca]) open possibilities that shallow noisy circuits do
not have, and their practical value depends on logical hardware, on
how the data are accessed, and on full resource estimates of the kind
the companion volume makes for its many-body algorithms.


## For many-body physics and quantum sensing

For the interests of the readers this book was written for, the
priority order above has a sharper form. We would prioritize
quantum process and Hamiltonian learning and adaptive experiment
design, and compare structured variational models against
measurement-based classical learning on the same physical problems.
The central question should be: *which quantum resource reduces
the number of experiments, improves the estimation accuracy, or lowers
the total computation at a fixed accuracy?* A proposal that cannot
name the resource and the quantity it improves is not yet a proposal.
The models of this book that pass the test are the ones whose data
are quantum: the kernel between quantum states, the Hamiltonian
learned from its Gibbs state, the thermal state prepared
variationally for a Hamiltonian of interest. The ones that do not,
yet, are the ones that load classical data into a circuit, and they
should be taught, as we have tried to teach them, alongside their
trainability limits, their classical surrogates and the quantum-data
alternatives. They remain useful tools; they are no longer a
sufficient research strategy on their own.


## Perspectives

The perspectives of the companion volume and of this book are the
same, seen from two sides, and we close with them.

**Algorithms.** 
On the near-term side: ansätze with the symmetries of the problem
built in, error mitigation by zero-noise extrapolation and
probabilistic error cancellation, remedies for barren plateaus by
locality, layer-wise training and informed initialization, and
benchmarks on tasks where the data are generated by quantum systems
in the first place. On the fault-tolerant side: Hamiltonian
simulation by Trotterization, qubitization and linear combinations of
unitaries, randomized and Bayesian phase estimation with fewer
qubits, quantum linear algebra read with its dequantizations, and
quantum walks and search as subroutines of Monte Carlo.

**Applications in the physical sciences.** 
Electronic structure, nuclear structure with the Lipkin, pairing and
shell-model Hamiltonians of the companion volume and of the many-body
text [MBbook], strongly correlated systems beyond the reach of
coupled-cluster theory, finite-temperature
properties by imaginary-time evolution and thermal-state preparation,
which may be the most natural application of the methods of
Section *Preparing Gibbs states on a quantum computer*; lattice gauge theories and the Hubbard
model, topological phases, and quantum sensing with
Heisenberg-limited parameter estimation, where the
learning-from-experiments results of Section *A priority list for the next years*
apply most
directly.

**Quantum machine learning.** 
Generative models beyond the restricted quantum Boltzmann machine,
quantum generative adversarial and diffusion models; quantum
reinforcement learning for quantum control and for adaptive
experiment design, the subject reserved for Chapter 6;
tensor networks and classical circuit simulation as both competitor
and guide; and classical shadow tomography and cross-entropy
benchmarking as the tools for certifying what a device has done.

**Hardware.** 
Superconducting, trapped-ion, photonic and neutral-atom platforms;
surface and color codes and the first logical qubits; and the open
question of how many logical qubits are needed for an advantage on a
problem anyone cares about. Quantum advantage requires better
algorithms *and* better hardware, in tandem, and the scale of
fault-tolerant machines will decide which of the algorithms of this
book are ever run on data too large to enumerate. The programs of the
preceding chapters are written so that readers can run them on
such a machine, or on its simulator, with the simulator of
`common/` replaced by a backend, and see for themselves whether
the picture changes.
